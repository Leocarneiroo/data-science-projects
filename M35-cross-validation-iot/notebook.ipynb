# 2 - Para essa base, onde você realizará as previsões de fire alarm, qual modelo de machine learning você aplicará? Justifique.




Para este problema, escolhi utilizar o Random Forest Classifier, já que a tarefa consiste em uma classificação binária: prever se houve incêndio (1) ou não (0). Como as variáveis preditoras são numéricas e contínuas, obtidas por sensores IoT, esse modelo se encaixa bem no contexto do problema.

A principal razão para essa escolha é que as variáveis ambientais — como TVOC, eCO2, material particulado, hidrogênio, etanol, pressão, temperatura e umidade — não se relacionam com a ocorrência de incêndio de forma simples ou linear. Em geral, o alerta não depende da variação de apenas um sensor isolado, mas da combinação de pequenas alterações em vários sensores ao mesmo tempo.

# 3 - Separe a base em Y e X e já rode a instância do modelo que você utilizará.

```python
x = base.drop('Fire Alarm', axis=1)
y = base['Fire Alarm']
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)
```

# 4 - Defina o número de Folds e rode o modelo com a validação cruzada.

```python
folds = 10
```

```python
crossvalidation = KFold(n_splits=folds, shuffle=True, random_state=5)
```

```python
modelo_final = cross_val_score(model, x, y, cv = folds)
```

```python
pontuacoes = cross_val_score(model, x, y, cv=crossvalidation)
print(f"Pontuações por fold: {pontuacoes}")
```

```python
print((modelo_final.mean()))
```
